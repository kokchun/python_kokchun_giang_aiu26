# matplotlib

```python
import numpy as np 

x = np.linspace(-2*np.pi, 2*np.pi,100)
x.shape
```

```python
y = np.sin(x)
```

## MATLAB style interface

```python
import matplotlib.pyplot as plt 

plt.figure(dpi=100, figsize=(12,4))
plt.plot(x,y)
plt.title("sin(x)")
plt.xlabel("x")
plt.ylabel("y")
```
