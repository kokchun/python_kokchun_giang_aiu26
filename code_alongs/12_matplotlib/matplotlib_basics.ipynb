# matplotlib

```python
import numpy as np 

x = np.linspace(-2*np.pi, 2*np.pi,100)
x.shape
```

```python
y = np.sin(x)
```

## MATLAB style interface - don't do this

```python
import matplotlib.pyplot as plt 

plt.figure(dpi=100, figsize=(12,4))
plt.plot(x,y)
plt.title("sin(x)")
plt.xlabel("x")
plt.ylabel("y")
plt.xlim([-3,3])
```

## OOP approach

```python
fig, ax = plt.subplots(1)
ax.plot(x, y)
ax.set(title="sin(x)", xlabel="x", ylabel="y", xlim=[-3, 3])
```

## subplots

```python
fig, ax = plt.subplots(1,2, figsize=(16,4))

ax[0].plot(x, np.sin(x))
ax[0].plot(x, np.cos(x))
ax[0].legend(["sin x", "cos x"])


ax[1].plot(x, np.exp(x))
ax[1].set(title="Exponential function")

fig.savefig("cool_graph.png", dpi=250)
```
