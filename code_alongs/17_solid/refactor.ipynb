# Refactor to follow SOLID

```python
from abc import ABC, abstractmethod

class Employee:
    def __init__(self, name, phone, salary, tax_rate):
        self.name = name
        self.phone = phone
        self.salary = salary
        self.tax_rate = tax_rate
    
    @property
    def net_salary(self):
        return self.salary * (1 - self.tax_rate)

class Report(ABC):
    @abstractmethod
    def content(self, employee: Employee) -> str: ...
```

## Note on abstract class

```python
class NotWorkingReport(Report):
    ...

try:
    NotWorkingReport()
except TypeError as err:
    print(err)
```

```python
class ExcelReport(Report):
    def content(self, employee):
        print("excel content")


ExcelReport()
```
