```python
import json

class Employee:
    def __init__(self, name, phone, salary):
        self.name = name
        self.phone = phone
        self.salary = salary
    
    # responsibility 1
    def net_salary(self, tax_rate=0.3):
        return self.salary * (1 - tax_rate)

    # responsibility 2
    # violate OCP - add pdf -> update this method
    # violate DIP - Employee depends on low lvl such as open and json
    def generate_report(self, report_type = "txt"):
        if report_type == "txt":
            with open(f"{self.name}.txt", "w") as file:
                file.write("Employee report\n")
                file.write(
                    f"Name: {self.name}\nPhone: {self.phone}\nSalary: {self.salary}\nNet salary: {self.net_salary()}"
                )
        elif report_type == "json":
            with open(f"{self.name}.json","w") as file:
                json.dump({
                    "name": self.name,
                    "phone": self.phone,
                    "salary": self.salary,
                    "net_salary": self.net_salary()
                }, file, indent=3)


e1 = Employee("Christer", "031-892121", 35000)
e1.generate_report(report_type="txt")
e1.generate_report(report_type="json")
```
