# Dictionary

```python
person = dict(
    name="Kokchun",
    age=35,
    work="teacher",
    interests=["math", "python", "linux"],
    phone=112,
)

person
```

```python
person = {
    "name": "Kokchun",
    "age": 35,
    "work": "teacher",
    "interests": ["math", "python", "linux"],
    "phone": 112,
}

person
```

```python
print(f"{person["name"] = }")
print(f"{person["age"] = }")
print(f"{person["interests"][1][-3:] = }")
```

## looping through a dict

```python
for key in person:
    print(f"{key}: {person[key]}")
```

```python
for key, value in person.items():
    print(f"{key}: {value}")
```
