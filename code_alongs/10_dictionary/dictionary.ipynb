# Dictionary

```python
person = dict(
    name="Kokchun",
    age=35,
    work="teacher",
    interests=["math", "python", "linux"],
    phone=112,
)

person
```

```python
person = {
    "name": "Kokchun",
    "age": 35,
    "work": "teacher",
    "interests": ["math", "python", "linux"],
    "phone": 112,
}

person
```

```python
print(f"{person["name"] = }")
print(f"{person["age"] = }")
print(f"{person["interests"][1][-3:] = }")
```

## looping through a dict

```python
for key in person:
    print(f"{key}: {person[key]}")
```

```python
for key, value in person.items():
    print(f"{key}: {value}")
```

```python
person["has_driving_license"] = True
person
```

```python
"ABCDEF"[::-1]
```

```python
# F<10, E:10, D:20, C:30, B:40, A:50

# empty dict
grade_limits = {}

for i, char_ in enumerate("ABCDEF"[::-1]):
    # assigns new keys and values to the dict
    grade_limits[char_] = i*10

grade_limits["F"] = "<10"
grade_limits
```

### printing out grade limits 

option 1

```python
# Betygsgränser
# A: 50
# B: 40
... 

grade_list = []

for grade, score in grade_limits.items():
    grade_list.append(f"{grade}: {score}")

print("Betygsgränser")
for grade_limit in grade_list[::-1]:
    print(grade_limit)
```

option 2

```python
print("Betygsgränser")
for key, value in reversed(grade_limits.items()):
    print(f"{key}: {value}")
```

option 3

```python

grades = sorted(grade_limits.keys())

for grade in grades:
    print(f"{grade}: {grade_limits[grade]}")
```

## dictionary comprehension

```python
grade_limits = {grade: i * 10 for i, grade in enumerate("FEDCBA")}
grade_limits
```
