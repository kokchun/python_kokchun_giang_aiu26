```python
v = np.array([1,2,5,-3,-2.5])
v
```

```python
print(f"{v[0] = }")
print(f"{v[2:] = }")
print(f"{v[-1] = }")
print(f"{v[::-1] = }")
```

## draw a graph

```python
import matplotlib.pyplot as plt
x = range(-3,4)
y = [-x**2 for x in x]

plt.plot(x,y)
```

```python
x = np.linspace(-3,3)
x
```
