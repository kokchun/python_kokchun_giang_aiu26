# while statement

- loops when condition is True
- "if-loop"

```python
i = 0

while i < 5:
    print(f"hi number {i}")
    i += 1 # increments i in each iteration
```

first 10 squares

```python
i = -10

while i < 10:
    # **2 to the power of 2
    square = i**2
    print(square, end="  ")
    i += 1
```
