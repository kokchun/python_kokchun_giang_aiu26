# while statement

- loops when condition is True
- "if-loop"

```python
i = 0

while i < 5:
    print(f"hi number {i}")
    i += 1 # increments i in each iteration
```

first 10 squares

```python
i = -10

while i <= 10:
    # **2 to the power of 2
    square = i**2
    print(square, end="  ")
    i += 1
```

## Oil leakage

There is an oil leakage causing the bird population in an island to 1/2 in each year. From start there were 80000 birds, how many whole years does it take for it to have less than 1/10 remaining?

```python
year = 0
while year < 10:
    year += 1

year
```

while statement good when you don't know in beforehand how many iterations you need

```python
year = 0
birds = 80000

while birds > 8000:
    print(f"year {year}, we have {birds} birds")
    birds = birds * 0.5
    year += 1

year
```

## break

```python
import random as rnd 


while True:
    dice = rnd.randint(1,6)

    guess = int(input("Guess the dice: "))

    if guess == dice:
        print("You WIN")
    else:
        print("You LOSE")

    play_again = input("Play again? y for yes")

    if play_again != "y":
        print("Have a nice day")
        break
```
