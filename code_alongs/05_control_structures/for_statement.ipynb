## random 

count number of fives in 1000 000 dice rolls -> calculate proportion of fives


```python
import random as rnd 

# type hinting
number_rolls: int = 10_000_000
count_fives: int = 0

for roll in range(number_rolls):
    dice = rnd.randint(1,6)

    if dice == 5:
        count_fives += 1

count_fives/number_rolls
```

## Loop through lists

```python
animals = ["ape", "rabbit", "dog", "fish", "cat"]

animals
```

```python
for animal in animals:
    print(f"cute animal: {animal}")
```

want this

```python
print(animals)
# common pattern similar to other languages
# don't do this
for i in range(len(animals)):
    print(f"{i}. {animals[i]}")
```

```python
# pythonic
for i, animal in enumerate(animals):
    print(f"{i}. {animal}")
```
