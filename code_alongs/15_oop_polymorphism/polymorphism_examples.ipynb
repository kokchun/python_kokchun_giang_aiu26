## Polymorphism class method example

```python
class Fish:
    def speak(self):
        print("blup blupp blupp")

    def __str__(self):
        return "fish swimming around"

class Fox:
    def speak(self):
        print("fox says")
    
    def __str__(self):
        return "Fox: no one know how I sound"

fish = Fish()
print(Fish)
print(fish)
```

```python
animals = (Fox(), Fox(), Fish(), Fish())
for animal in animals:
    #print(animal)
    animal.speak()
```

## Geometry

```python
import numpy as np


class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return self.radius**2 * np.pi


class Square:
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side**2


class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def perimeter(self):
        return self.width * 2 + self.height * 2

    def area(self):
        return self.width * self.height


# circle = Circle(1)
# square = Square(2)

# shapes = [circle, square]

shapes = [Circle(1), Square(2), Rectangle(3,5), Rectangle(1,5)]
area = 0

for shape in shapes:
    area += shape.area()

area
```
