# Polymorphism

- many morphs (forms)
- a function/method have different abilities based on their type

## Examples

```python
# plus operator -> concatenation 
"35"+"3"
```

```python
[2,5,3] + [1,2]
```

```python
try:
    [2,5,3] + (1,2)
except TypeError as err:
    print(err)

try:
    5+"5"
except TypeError as err:
    print(err)


# multiplication operator -> different behaviors for different types
5*"5", 5*5, [2]*5, (2,)*5 
```

```python
import numpy as np 

np.ones(10)*5
```
