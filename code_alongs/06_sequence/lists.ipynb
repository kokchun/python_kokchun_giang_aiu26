# lists

```python
cities = ["göteborg", "varberg", "falkenberg", "stockholm", "malmö", "halmstad"]
print(f"{cities[0] = }")
print(f"{cities[3] = }")
print(f"{cities[-1] = }")
print(f"{cities[-3] = }")

# a string 'stockholm' is a list of characters 
print(f"{cities[-3][0] = }")
print(f"{cities[-3][5] = }")
```
