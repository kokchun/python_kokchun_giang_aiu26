# lists

accessing list

```python
cities = ["göteborg", "varberg", "falkenberg", "stockholm", "malmö", "halmstad"]
print(f"{cities[0] = }")
print(f"{cities[3] = }")
print(f"{cities[-1] = }")
print(f"{cities[-3] = }")

# a string 'stockholm' is a list of characters 
print(f"{cities[-3][0] = }")
print(f"{cities[-3][5] = }")
```

slicing a list

```python
print(cities)
# slicing a list
print(f"{cities[1:4] = }")
print(f"{cities[3:7] = }")
print(f"{cities[3:] = }")
```

modify list

```python
# mutate list
cities[3] = "kungsbacka"
cities
```

sort a list

```python
cities.sort()
cities
```

```python
numbers = [-1,-5,5,9,3]
numbers.sort(reverse=True)

numbers
```

```python
greek_letters = ["beta", "gamma", "alpha", "epsilon"]
greek_letters = sorted(greek_letters)
greek_letters
```

## loop through list

```python
for city in cities:
    print(city)
```

```python
more_cities = ["stockholm", "örebro", "umeå", "luleå"]

for city in more_cities:
    cities.append(city)

cities
```

create a list of cubes between 0 - 10

```python
cubes = []

for i in range(11):
    cubes.append(i**3)

cubes
```
