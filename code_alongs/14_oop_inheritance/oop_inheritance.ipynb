# OOP inheritance

## Properties


### Without properties


```python
class Person:
    def __init__(self, name: str, age: int | float):
        self.age = age
        self.name = name
    
    def __repr__(self):
        return f"Person('{self.name}', {self.age})"

p1 = Person("Pia", 50)
p1
```

```python
p2 = Person("Fia", "tjugo")
p2
```

### With properties

```python
class Person:
    def __init__(self, name: str, age: int | float):
        self.age = age
        self.name = name
    
    def __repr__(self):
        return f"Person('{self.name}', {self.age})"
```
