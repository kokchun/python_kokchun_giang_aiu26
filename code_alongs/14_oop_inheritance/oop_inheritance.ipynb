# OOP inheritance

## Properties


### Without properties


```python
class Person:
    def __init__(self, name: str, age: int | float):
        self.age = age
        self.name = name
    
    def __repr__(self):
        return f"Person('{self.name}', {self.age})"

p1 = Person("Pia", 50)
p1
```

```python
p2 = Person("Fia", "tjugo")
p2
```

### With properties

```python
class Person:
    def __init__(self, name: str, age: int | float):
        self.age = age
        self.name = name

    @property
    def age(self) -> int|float:
        print("getter is running ...")
        return self._age

    @age.setter
    def age(self, value: int|float):
        print("setter is running ...")
        # error handling
        if not isinstance(value, (int, float)):
            raise TypeError(f"Age must be int or float not {type(value)}")
        self._age = value


    def __repr__(self):
        return f"Person('{self.name}', {self.age})"

# setter runs when we assign a value to a property
p3 = Person("Marcus", 15)

# getter runs when we access the property
p3.age
```

```python
p3.age = 20
p3
```
