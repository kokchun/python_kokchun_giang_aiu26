# OOP inheritance

## Properties


### Without properties


```python
class Person:
    def __init__(self, name: str, age: int | float):
        self.age = age
        self.name = name
    
    def __repr__(self):
        return f"Person('{self.name}', {self.age})"

p1 = Person("Pia", 50)
p1
```

```python
p2 = Person("Fia", "tjugo")
p2
```

### With properties

```python
class Person:
    def __init__(self, name: str, age: int | float):
        self.age = age
        self._name = name

    # decorator -> gives extra functionality to a function
    @property
    def age(self) -> int|float:
        # print("getter is running ...")
        return self._age

    @age.setter
    def age(self, value: int|float):
        # print("setter is running ...")
        # error handling
        if not isinstance(value, (int, float)):
            raise TypeError(f"Age must be int or float not {type(value)}")

        if not 0 <= value < 125:
            raise ValueError(f"Age must be between 0 and 124 not {value}")
        
        # backing variable
        self._age = value

    # read-only property <- have not specified a setter 
    @property
    def name(self):
        return self._name

    def __repr__(self):
        return f"Person('{self.name}', {self.age})"

# setter runs when we assign a value to a property
p3 = Person("Marcus", 15)

try:
    p3.age = -5
except ValueError as err:
    print(err)

try:
    p3.name = "hej"
except AttributeError as err:
    print(err)

p3.age = 5
p3.age
```

### Inheritance and composition

```python
class Student(Person):
    def __init__(self, name, age, gpa = 0):
        super().__init__(name, age)

        self.gpa = gpa

    # override the __repr__ from base class
    def __repr__(self):
        return f"Student('{self.name}', {self.age}, {self.gpa})"


s1 = Student('Doddo', 4, 2)
try: 
    s1.age = -5
except ValueError as err:
    print(err)

s1
```
