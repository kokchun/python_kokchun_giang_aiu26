# Encapsulation, "private", property

encapsulation
- hide implementation details
- property -> getter and setter

private
- don't access this attribute/method from outside
- private by convention in Python _ or __ as prefix





## private

```python
from datetime import datetime

class Student:
    def __init__(self, name, program, personal_number, csn = True):
        self.name = name
        self.program = program
        # private by convention 
        self._personal_number = personal_number

    def age(self):
        # hide info on implementation
        current_year = datetime.now().year

        # ... also add logic on 20 or 19
        birth_year = int("20"+self._personal_number[:2])

        age = current_year - birth_year

        return age
         

    def __str__(self):
        return f"Student {self.name} is {self.age()} age years and studies {self.program}"

    def __repr__(self):
        return f"Student('{self.name}', '{self.program}')"    

s1 = Student("Babba", "AI26", "020302-9988")
print(s1.age())
s1
```
